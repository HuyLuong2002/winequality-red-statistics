```python
import pandas as pd
from pathlib import Path

# Đường dẫn tập dữ liệu
DATA_PATH = Path("dataset/winequality-red.csv")

# Đọc dữ liệu
df = pd.read_csv(DATA_PATH)
```

## Nguồn dữ liệu

```python
nguon_du_lieu = {
    "Tên tập dữ liệu": "Wine Quality Red",
    "Nguồn": "UCI Machine Learning Repository",
    "URL": "https://archive.ics.uci.edu/ml/datasets/wine+quality",
    "Mô tả": "Các thuộc tính hóa học của rượu vang đỏ vùng Vinho Verde (Bồ Đào Nha) và điểm chất lượng do chuyên gia đánh giá.",
    "File cục bộ": str(DATA_PATH.resolve()),
}

for key, value in nguon_du_lieu.items():
    print(f"• {key}: {value}")
```

## Số lượng mẫu

```python
so_mau = len(df)

print(f"• Số lượng mẫu (số dòng): {so_mau:,}")
print(f"• Số thuộc tính (số cột): {df.shape[1]}")
print(f"• Kích thước tập dữ liệu: {df.shape[0]:,} dòng × {df.shape[1]} cột")
```

## Các thuộc tính

```python
print("Danh sách các thuộc tính:")
for i, col in enumerate(df.columns, start=1):
    print(f"  {i}. {col}")
```

## Kiểu dữ liệu của từng thuộc tính

```python
kieu_du_lieu = pd.DataFrame({
    "Thuộc tính": df.columns,
    "Kiểu dữ liệu": df.dtypes.astype(str).values,
})

display(kieu_du_lieu)
```

## Tóm tắt

```python
print("=" * 50)
print("TÓM TẮT MÔ TẢ TẬP DỮ LIỆU")
print("=" * 50)
print(f"\n1. Nguồn dữ liệu: UCI ML Repository - Wine Quality Red")
print(f"2. Số lượng mẫu: {so_mau:,}")
print(f"3. Số thuộc tính: {len(df.columns)}")
print(f"4. Kiểu dữ liệu:")
print("\n" + df.dtypes.to_string())
```
